zadanie 10

```python
dataset_a = {'wiek', 'plec', 'miasto', 'dochod', 'wyksztalcenie', 'stan_cywilny'}
dataset_b = {'wiek', 'plec', 'miasto', 'pensja', 'stanowisko', 'staz'}

print("cechy wspolne dataset_a & dataset_b:", dataset_a & dataset_b) 
print("cechy tylko w dataset_a:", dataset_a - dataset_b) 
print("cechy tylko w dataset_b:", dataset_b - dataset_a) 
print("cechy unikalne:", dataset_a | dataset_b) 
print("roznica symetryczna :", dataset_a ^ dataset_b) 
print({'wiek', 'plec'}.issubset(dataset_a | dataset_b))
```

```python
sprzedaz = {
    (2024, 'Q1'): 120_000,
    (2024, 'Q2'): 145_000,
    (2024, 'Q3'): 138_000,
    (2024, 'Q4'): 167_000,
    (2025, 'Q1'): 130_000,
    (2025, 'Q2'): 155_000,
    (2025, 'Q3'): 149_000
}


print("calkowita",sum(sprzedaz.values()))
print("min",min(sprzedaz.values()))
print("max",max(sprzedaz.values()))


klucz_max = max(sprzedaz, key=sprzedaz.get)
rok_max, kw_max = klucz_max
print(f"Najwyższa sprzedaż: {rok_max} {kw_max} – {sprzedaz[klucz_max]}.")

posortowane_klucze = sorted(sprzedaz, key=sprzedaz.get, reverse=True)
print(posortowane_klucze[0])
print(posortowane_klucze[1])
print(posortowane_klucze[2])

lata = set()
for klucz in sprzedaz.keys():
    rok = klucz[0]  
    lata.add(rok)   

print("unikalne lata z kluczy",lata)
```
