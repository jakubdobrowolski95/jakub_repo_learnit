```python
values = [75, 50, 100, 62.5]
mix = 50
max = 100
normalized = [(value - mix) / (max - mix) for value in values]
```

```python
for normalize in normalized:
    if 0 <= normalize <= 1:
        print(f"{normalize} is normalized")
    else:
        print(f"{normalize} is not normalized")
```

```python
import pandas as pd

tabela_normalizacji = pd.DataFrame({
    "before_normalized": values,
    "after_normalized": normalized,
})

tabela_normalizacji
```

```python
import math

samples = 50000
batch_size = 128
epochs = 50
time_per_batch = 0.15

# Liczba batchy na epokę
batches_per_epoch = math.ceil(samples / batch_size)

# Całkowity czas w sekundach
total_seconds = batches_per_epoch * epochs * time_per_batch

# Zamiana na godziny, minuty i sekundy
hours = int(total_seconds // 3600)
minutes = int((total_seconds % 3600) // 60)
seconds = int(total_seconds % 60)

print(f"szacowany czas to {hours} godzin, {minutes} minutt i {seconds} sekund")

```

zadanie 15


```python
import statistics 

weights = [65.5, 78.2, 54.8, 92.1, 70.0]
avg = statistics.mean(weights)
std_dev = statistics.stdev(weights)
print(avg)
print(std_dev)
```

```python
z_score = [(weight - avg) / std_dev for weight in weights]
print(z_score)
```

```python
for z in z_score:
    if z > 2:
        print(f"Z-SCORE {z} jest wysokie")
    elif abs(z) < 1:
        print(f"Z-SCORE {z} typowa")
    elif z < -2 :
        print(f"Z-SCORE {z} bardzo niska")
```
