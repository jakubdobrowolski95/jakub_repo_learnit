```python
class Produkt:
    def __init__(self, nazwa, cena, ilosc_w_magazynie):
        self.nazwa = nazwa
        self.cena = cena
        self.ilosc_w_magazynie = ilosc_w_magazynie
    def pokaz_info(self):
        print(f"Nazwa produktu {self.nazwa}, cena {self.cena}")
    def sprzedaj(self, ilosc):
        if ilosc > self.ilosc_w_magazynie:
            print("Brak wystarczającej ilości w magazynie")
            return
        else:
            self.ilosc_w_magazynie -= ilosc
            print(f"Zostalo w magazynie {self.ilosc_w_magazynie}")
    def dostawa(self, ilosc):
         print(f"Dostawa {ilosc} ")
         self.ilosc_w_magazynie += ilosc

        
produkt_1 = Produkt("banan", 2, 5)
produkt_2 = Produkt("banan1", 2, 5)
produkt_3 = Produkt("banan2", 2, 5)
produkt_1.pokaz_info()
produkt_1.sprzedaj(4)
produkt_1.dostawa(5)
```

```python

class Kalkulator:
    def __init__(self, a, b):
        self.a = a 
        self.b = b
        self.wynik = 0
    def dodaj(self):
        self.wynik = self.a + self.b 
        print(f"wynik to {self.wynik}")

    def odejmij(self):
        self.wynik = self.a - self.b
        print(f"wynik to {self.wynik}")
    def pomnoz(self):
        self.wynik = self.a * self.b
        print(f"wynik to {self.wynik}")

    def podziel(self):
        if self.b == 0:
            print("Nie można dzielić przez zero")
            return
        self.wynik = self.a / self.b
        print(f"wynik to {self.wynik}")

kalkulacja = Kalkulator(2,5)
kalkulacja.dodaj()
kalkulacja.odejmij()
kalkulacja.pomnoz()
kalkulacja.podziel()
```

```python
class BankAccount:
    def __init__(self, wlasciciel, saldo=0):
        self.wlasciciel = wlasciciel
        self.saldo = saldo

    def wplac(self, kwota):
        if kwota <= 0:
            print("Nie mozna wplacac ujemnych wartosci")
        else:
            self.saldo += kwota
            print(f"Wplaciles {kwota}, masz teraz {self.saldo}")

    def wyplac(self, kwota):
        if kwota <= 0:
            print("Nie mozna wyplacac ujemnych wartosci")
        elif kwota > self.saldo:
            print("Brak wystarczajacych srodkow")
        else:
            self.saldo -= kwota
            print(f"Wyplaciles {kwota}, masz teraz {self.saldo}")

    def pokaz_saldo(self):
        print(f"{self.wlasciciel}: saldo = {self.saldo}")


konto = BankAccount("jakub")
konto.wplac(1)      # błąd, ujemna kwota
konto.wplac(10)      # 15
konto.wyplac(100)    # brak środków
konto.wyplac(5)      # 10
konto.pokaz_saldo()
```

```python
class Ksiazka:
    def __init__(self, tytul, autor, rok_wydania, liczba_stron):
        self.tytul = tytul
        self.autor = autor
        self.rok_wydania = rok_wydania
        self.liczba_stron = liczba_stron

    def __str__(self):
        return f"Ksiazka '{self.tytul}' napisana przez {self.autor}, wydana w roku {self.rok_wydania}, ma {self.liczba_stron} stron"


ksiazka1 = Ksiazka("Ferdydurka", "Witold Gombrowicz", 1936, 215)
ksiazka2 = Ksiazka("Lalka", "Boleslaw Prus", 1890, 600)
ksiazka3 = Ksiazka("Pan Tadeusz", "Adam Mickiewicz", 1834, 300)
ksiazka4 = Ksiazka("Quo Vadis", "Henryk Sienkiewicz", 1896, 450)
ksiazka5 = Ksiazka("Wesele", "Stanislaw Wyspianski", 1901, 150)

ksiazki = [ksiazka1, ksiazka2, ksiazka3, ksiazka4, ksiazka5]

for ksiazka in ksiazki:
    print(ksiazka)
```

```python
class Student:
    def __init__(self, imie, numer_indeksu):
        self.imie = imie
        self.numer_indeksu = numer_indeksu
        self.oceny = []

    def dodaj_ocene(self, ocena):
        self.oceny.append(ocena)

    def srednia(self):
        if len(self.oceny) == 0:
            return 0
        return sum(self.oceny) / len(self.oceny)

    def __str__(self):
        return f"{self.imie} (indeks {self.numer_indeksu}) - oceny: {self.oceny}, srednia: {self.srednia():.2f}"


student1 = Student("Jakub", "12345")
student1.dodaj_ocene(5)
student1.dodaj_ocene(4)
student1.dodaj_ocene(3)
print(student1)


```

```python
class Pojazd:
    def __init__(self, marka, model, rok):
        self.marka = marka
        self.model = model
        self.rok = rok

    def pokaz_info(self):
        print(f"Pojazd: {self.marka} {self.model} z roku {self.rok}")


class Samochod(Pojazd):
    def __init__(self, marka, model, rok, liczba_drzwi):
        super().__init__(marka, model, rok)
        self.liczba_drzwi = liczba_drzwi

    def pokaz_info(self):
        print(f"Samochod: {self.marka} {self.model} z roku {self.rok}, liczba drzwi: {self.liczba_drzwi}")


pojazd1 = Samochod("Toyota", "Corolla", 444, 4)
pojazd2 = Pojazd("Fiat", "126p", 555)

pojazd1.pokaz_info()
pojazd2.pokaz_info()
```

```python
class Pracownik:
    def __init__(self, imie, stanowisko, pensja):
        self.imie = imie
        self.stanowisko = stanowisko
        self.pensja = pensja

    def podwyzka(self, procent):
        if procent <= 0:
            print("Podwyzka nie moze byc rowna lub mniejsza 0")
        else:
            self.pensja = self.pensja + self.pensja * (procent / 100)
            print(f"Nowa pensja to {self.pensja} po podwyzce {procent}%")

    def __str__(self):
        return f"Pracownik {self.imie} ({self.stanowisko}), pensja: {self.pensja}"


pracownik1 = Pracownik("Jakub", "Analityk", 5000)
pracownik2 = Pracownik("Anna", "Programista", 7000)
pracownik3 = Pracownik("Piotr", "Manager", 9000)

pracownik1.podwyzka(5)
pracownik2.podwyzka(10)
pracownik3.podwyzka(3)

print(pracownik1)
print(pracownik2)
print(pracownik3)
```

```python
class Zadanie:
    def __init__(self, opis, status="do zrobienia"):
        self.opis = opis
        self.status = status

    def oznacz_jako_zrobione(self):
        self.status = "zrobione"

    def __str__(self):
        return f"{self.opis} - status: {self.status}"


zadanie1 = Zadanie("Wynies smieci")
zadanie2 = Zadanie("Zrob zakupy")
zadanie3 = Zadanie("Napisz raport")
zadanie4 = Zadanie("Umyj naczynia")
zadanie5 = Zadanie("Przeczytaj ksiazke")

zadanie1.oznacz_jako_zrobione()
zadanie3.oznacz_jako_zrobione()

print(zadanie1)
print(zadanie2)
print(zadanie3)
print(zadanie4)
print(zadanie5)
```

```python
class FiltrowanieDodatnich:
    def process(self, dane):
        return [x for x in dane if x >= 0]
class Zaokraglanie:
    def process(self, dane):
        return [round(x) for x in dane]
class Skalowanie:
    def __init__(self, wspolczynnik):
        self.wspolczynnik = wspolczynnik

    def process(self, dane):
        return [x * self.wspolczynnik for x in dane]


class Sumowanie:
    def process(self, dane):
        return sum(dane)


class Pipeline:
    def __init__(self, kroki):
        self.kroki = kroki

    def run(self, dane):
        for krok in self.kroki:
            dane = krok.process(dane)
        return dane


dane = [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]

pipeline = Pipeline([
    FiltrowanieDodatnich(),
    Zaokraglanie(),
    Skalowanie(2),
    Sumowanie(),
])

print(pipeline.run(dane))
```

```python
class Akcja:
    def __init__(self, nazwa, cena_kupna, cena_aktualna):
        self.nazwa = nazwa
        self.cena_kupna = cena_kupna
        self.cena_aktualna = cena_aktualna

    def __repr__(self):
        return f"{self.nazwa} (kupno: {self.cena_kupna}, aktualna: {self.cena_aktualna})"


class Portfel:
    def __init__(self, kapital_poczatkowy):
        self.akcje = [] 
        self.kapital_poczatkowy = kapital_poczatkowy
        self.gotowka = kapital_poczatkowy

    def kup_akcje(self, akcja, ilosc):
        koszt = akcja.cena_kupna * ilosc
        if koszt > self.gotowka:
            print(f"Brak środków, aby kupić {ilosc} x {akcja.nazwa} (koszt: {koszt}, dostępne: {self.gotowka})")
            return

        self.gotowka -= koszt

  
        for i, (a, ile) in enumerate(self.akcje):
            if a.nazwa == akcja.nazwa:
                self.akcje[i] = (a, ile + ilosc)
                print(f"Dokupiono {ilosc} x {akcja.nazwa}, łącznie: {ile + ilosc}")
                return

        self.akcje.append((akcja, ilosc))
        print(f"Kupiono {ilosc} x {akcja.nazwa} za {koszt}")

    def sprzedaj_akcje(self, akcja, ilosc):
        for i, (a, ile) in enumerate(self.akcje):
            if a.nazwa == akcja.nazwa:
                if ilosc > ile:
                    print(f"Nie można sprzedać {ilosc} x {a.nazwa}, posiadasz tylko {ile}")
                    return

                przychod = a.cena_aktualna * ilosc
                self.gotowka += przychod

                if ilosc == ile:
                    self.akcje.pop(i)
                else:
                    self.akcje[i] = (a, ile - ilosc)

                print(f"Sprzedano {ilosc} x {a.nazwa} za {przychod}")
                return

        print(f"Nie posiadasz akcji {akcja.nazwa}")

    def aktualizuj_ceny(self, slownik_cen):
        for a, ilosc in self.akcje:
            if a.nazwa in slownik_cen:
                a.cena_aktualna = slownik_cen[a.nazwa]

    def wartosc_portfela(self):
        wartosc_akcji = sum(a.cena_aktualna * ilosc for a, ilosc in self.akcje)
        return wartosc_akcji + self.gotowka

    def zysk_strata(self):
        return self.wartosc_portfela() - self.kapital_poczatkowy

    def pokaz_stan(self):
        print(f"\nGotówka: {self.gotowka:.2f}")
        print("Posiadane akcje:")
        for a, ilosc in self.akcje:
            print(f"  {a.nazwa}: {ilosc} szt. @ {a.cena_aktualna} = {a.cena_aktualna * ilosc:.2f}")
        print(f"Wartość portfela: {self.wartosc_portfela():.2f}")
        print(f"Zysk/strata: {self.zysk_strata():.2f}")




if __name__ == "__main__":
    portfel = Portfel(kapital_poczatkowy=10000)


    apple = Akcja("AAPL", cena_kupna=150, cena_aktualna=150)
    tesla = Akcja("TSLA", cena_kupna=250, cena_aktualna=250)
    google = Akcja("GOOGL", cena_kupna=100, cena_aktualna=100)
    amazon = Akcja("AMZN", cena_kupna=120, cena_aktualna=120)
    nvidia = Akcja("NVDA", cena_kupna=400, cena_aktualna=400)

    print("=== KUPNO AKCJI ===")
    portfel.kup_akcje(apple, 10)   
    portfel.kup_akcje(tesla, 5)    
    portfel.kup_akcje(google, 8)  
    portfel.kup_akcje(amazon, 6)   
    portfel.kup_akcje(nvidia, 3)    

    portfel.pokaz_stan()

    print("\n=== AKTUALIZACJA CEN (zmiany rynkowe) ===")
    nowe_ceny = {
        "AAPL": 165,  
        "TSLA": 200, 
        "GOOGL": 110,
        "AMZN": 100,  
        "NVDA": 480,   
    }
    portfel.aktualizuj_ceny(nowe_ceny)
    portfel.pokaz_stan()

    print("\n=== CZĘŚCIOWA SPRZEDAŻ ===")
    portfel.sprzedaj_akcje(tesla, 5)    
    portfel.sprzedaj_akcje(nvidia, 1)  

    portfel.pokaz_stan()

    print("\n=== PRÓBA SPRZEDAŻY ZA DUŻO ===")
    portfel.sprzedaj_akcje(amazon, 100)  
```
