```python
np.random.seed(42)
data = np.concatenate([np.random.randn(100) * 10 + 50,
np.array([150, -50, 200, -100, 180])
])
np.random.shuffle(data)
```

```python
data
```

```python
srednia = np.mean(data)
srednia
```

```python
odchylenie_std = np.std(data)
odchylenie_std
```

```python
z_score = (data - srednia) / odchylenie_std
z_score
```

```python
outliers = np.abs(z_score) <= 3
```

```python
data_bez_outlierow = data[outliers]
print(data_bez_outlierow)
```

```python
odchylenie_std_bez_outlierow =np.std(data_bez_outlierow)
print(f"Odchylenie standardowe przed usunięciem outlierów: {odchylenie_std}")
print(f"Odchylenie standardowe po usunięciu outlierów: {odchylenie_std_bez_outlierow}")
```

```python
sredina_data_bez_outlierow = np.mean(data_bez_outlierow)
min_data_bez_outlierow = np.min(data_bez_outlierow)
max_data_bez_outlierow = np.max(data_bez_outlierow)
liczba_elementow_bez_outlierow = len(data_bez_outlierow)
print(f"Średnia przed usunieciem outlierow to: {srednia}, a po to {sredina_data_bez_outlierow}")
print(f"Min przed usunieciem outlierow to: {np.min(data)}, a po to {min_data_bez_outlierow}")
print(f"Max przed usunieciem outlierow to: {np.max(data)}, a po to {max_data_bez_outlierow}")
print(f"Liczba elementow przed usunieciem outlierow to: {len(data)}, a po to {liczba_elementow_bez_outlierow}") 
```

```python

def tekstowy_histogram(dane, liczba_binow=10, szerokosc=50, tytul=""):
    licznosci, krawedzie = np.histogram(dane, bins=liczba_binow)
    max_licznosc = np.max(licznosci)
    
    print(f"\n{tytul}")
    print("-" * 60)
    for i in range(len(licznosci)):
        przedzial = f"{krawedzie[i]:6.1f} - {krawedzie[i+1]:6.1f}"
        dlugosc_slupka = int((licznosci[i] / max_licznosc) * szerokosc)
        slupek = "█" * dlugosc_slupka
        print(f"{przedzial} | {slupek} {licznosci[i]}")
tekstowy_histogram(data, liczba_binow=10, tytul="Dane ORYGINALNE (z outlierami)")
tekstowy_histogram(data_bez_outlierow, liczba_binow=10, tytul="Dane BEZ outlierów")
```
