```python
def narysuj_wektory_2d(vektory_z_etykietami, xlim=(-1, 5), ylim=(-1, 5)):
    x_min, x_max = xlim
    y_min, y_max = ylim
    width = x_max - x_min + 1
    height = y_max - y_min + 1

    # Siatka (pusta)
    grid = [[' .' for _ in range(width)] for _ in range(height)]

    # Oś X i Y
    for row in range(height):
        y_val = y_max - row
        if y_val == 0:
            grid[row] = [' -' for _ in range(width)]

    for col in range(width):
        x_val = x_min + col
        for row in range(height):
            if x_val == 0:
                grid[row][col] = ' |' if grid[row][col] != ' -' else ' +'

    # Umieść symbol w punkcie (x, y)
    for vec, symbol in vektory_z_etykietami:
        x, y_c = int(vec[0]), int(vec[1])
        col = x - x_min
        row = y_max - y_c
        if 0 <= col < width and 0 <= row < height:
            grid[row][col] = f' {symbol}'

    print(f" y={y_max}" + ' ' * (width * 2 - 5) + f"y={y_min}")
    for row in range(height):
        y_val = y_max - row
        prefix = f"y={y_val:2d} |"
        print(prefix + ''.join(grid[row]))
    print("     " + ''.join(f"{x_min + c:2d}" for c in range(width)))
    print(f" (x={x_min} .. x={x_max})")


narysuj_wektory_2d([(v1, 'A'), (v2, 'B'), (v_suma, 'S')])
```

```python
import numpy as np

def make_synthetic_flowers(seed=42):
    """Generuje syntetyczny dataset kwiatów (150 x 4) z 3 gatunkami."""
    rng = np.random.default_rng(seed)
    n_per_class = 50
    centers = np.array([
        [5.0, 3.4, 1.5, 0.2],
        [5.9, 2.8, 4.3, 1.3],
        [6.6, 3.0, 5.6, 2.0],
    ])
    stds = np.array([0.35, 0.35, 0.45, 0.25])
    X_list, y_list = [], []
    for class_idx, center in enumerate(centers):
        samples = rng.normal(loc=center, scale=stds, size=(n_per_class, 4))
        X_list.append(samples)
        y_list.append(np.full(n_per_class, class_idx))
    X = np.vstack(X_list)
    y = np.concatenate(y_list)
    feature_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
    target_names = np.array(['gatunek_A', 'gatunek_B', 'gatunek_C'])
    return X, y, feature_names, target_names



X, y, feature_names, target_names = make_synthetic_flowers(seed=42)


petal_length = X[:, 2]


min_val = petal_length.min()
max_val = petal_length.max()

petal_length_norm = (petal_length - min_val) / (max_val - min_val)

print(f"Min: {min_val:.4f}, Max: {max_val:.4f}\n")
print("Przed normalizacją (pierwsze 10):")
print(np.round(petal_length[:10], 4))

print("\nPo normalizacji (pierwsze 10):")
print(np.round(petal_length_norm[:10], 4))

print(f"\nMin po normalizacji: {petal_length_norm.min():.6f}")
print(f"Max po normalizacji: {petal_length_norm.max():.6f}")
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
iloczyn_skalarny = a @ b
norma_a = np.linalg.norm(a)
norma_b = np.linalg.norm(b)
cos_theta = iloczyn_skalarny / (norma_a * norma_b)
cos_theta
kat_rad = np.arccos(cos_theta)
kat_deg = np.degrees(kat_rad)

print(f"Iloczyn skalarny (a·b): {iloczyn_skalarny}")
print(f"Norma a: {norma_a:.2f}")
print(f"Norma b: {norma_b:.2f}")
print(f"cos(theta): {cos_theta:.2f}")
print(f"Kąt w radianach: {kat_rad:.2f}")
print(f"Kąt w stopniach: {kat_deg:.2f}")
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
```

```python
fare_pcclass = df[['Fare','Pclass']]
fare_pcclass = fare_pcclass.dropna().reset_index(drop=True)
```

```python

fare = fare_pcclass['Fare'].to_numpy()
pclass = fare_pcclass['Pclass'].to_numpy()

group_means = np.empty(len(fare))
group_stds = np.empty(len(fare))

for c in [1, 2, 3]:
    mask = pclass == c
    
    mean = fare[mask].mean()
    std = fare[mask].std(ddof=1)
    
    group_means[mask] = mean
    group_stds[mask] = std

fare_normalized_np = (fare - group_means) / group_stds
np.round(fare_normalized_np, 6)
```

```python
fare_pcclass = df[['Fare', 'Pclass']].dropna()

fare_pcclass['mean_fare'] = (
    fare_pcclass.groupby('Pclass')['Fare'].transform('mean')
)

fare_pcclass['std_fare'] = (
    fare_pcclass.groupby('Pclass')['Fare'].transform('std')
)

fare_pcclass['fare_normalized_pd'] = (
    fare_pcclass['Fare'] - fare_pcclass['mean_fare']
) / fare_pcclass['std_fare']

fare_pcclass
```
