```python
onechunk = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv', chunksize=50)
pierwszy_chunk = next(onechunk)
```

```python
total_passengers = 0
age_sum = 0
age_count = 0
survived_sum = 0

for chunk in pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv', chunksize=50):
    total_passengers += len(chunk) 
    age_sum += chunk['Age'].sum()
    age_count += chunk['Age'].notnull().sum()
    survived_sum += chunk['Survived'].sum()

avg_age = age_sum / age_count 
survival_rate = (survived_sum / total_passengers) * 100

```

```python
full_df_total_passengers = len(df10)
avg_age_full_df = df10['Age'].mean()
survival_rate_full_df = (df10['Survived'].sum() / full_df_total_passengers) * 100
```

```python
print(f"Średni wiek pasażerów (przetwarzanie w kawałkach): {avg_age}")
print(f"Średni wiek pasażerów (pełny DataFrame): {avg_age_full_df}")
print(f"calkowita liczba pasażerów (przetwarzanie w kawałkach): {total_passengers}")
print(f"calkowita liczba pasażerów (pełny DataFrame): {full_df_total_passengers}")
print(f"survival rate (przetwarzanie w kawałkach): {survival_rate}%")
print(f"survival rate (pełny DataFrame): {survival_rate_full_df}%")
```
